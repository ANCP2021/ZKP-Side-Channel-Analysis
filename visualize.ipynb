```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
def load_dstat_file(filepath):
    df = pd.read_csv(filepath, skiprows=5)
    
    if df.columns[-1] != 'time':
        df.columns = list(df.columns[:-1]) + ['time']
    
    return df
```

```python
def plot_sys_usage(df, filename_suffix):
    metric_name = 'sys'
    
    seconds = np.arange(len(df))
    
    plt.figure(figsize=(8, 6))
    plt.plot(seconds, df[metric_name], linewidth=2)
    
    # Axis labels
    plt.xlabel('Seconds Elapsed', fontsize=18)
    plt.ylabel('CPU Usage (%)', fontsize=18)
    
    # Set font size for ticks
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    
    # Y-axis: CPU Usage (%)
    plt.ylim(0, 10)  # 0-10 + little buffer
    
    # X-axis: Seconds
    if 'lambda' in filename_suffix.lower():
        plt.xlim(0, 35)
    elif 'pi' in filename_suffix.lower():
        plt.xlim(0, 45)
    else:
        raise ValueError(f"Cannot determine if file is 'lambda' or 'pi' from filename: {filename_suffix}")
    
    plt.grid(True)
    plt.tight_layout()
    
    # Save to PDF
    output_filename = f'cpu_usage_sys_{filename_suffix}.pdf'
    plt.savefig(output_filename)
    plt.show()
    plt.close()
```

```python
snark_lambda = load_dstat_file('SNARK_lambda_stat.md')
snark_pi = load_dstat_file('SNARK_pi_stat.md')
stark_lambda = load_dstat_file('STARK_lambda_stat.md')
stark_pi = load_dstat_file('STARK_pi_stat.md')

# plot_sys_usage(snark_lambda, 'snark_lambda')
# plot_sys_usage(snark_pi, 'snark_pi')
# plot_sys_usage(stark_lambda, 'stark_lambda')
plot_sys_usage(stark_pi, 'stark_pi')
```
